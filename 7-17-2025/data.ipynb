```python
import pandas as pd
import matplotlib.pyplot as plt
import json
```

```python
df=pd.read_json('mythical.json')
df
```

```python
with open ('mythical.json') as file:
    data=json.load(file)
```

```python
mythicals_list = data['mythicals']
mythicals_list
```

```python
datafm=pd.DataFrame(mythicals_list)
datafm
```

```python
creatures=datafm['creature']
creatures
```

```python
plt.title("creatures sightings")
plt.plot(datafm['creature'],datafm['year'])
plt.xlabel("monsters")
plt.ylabel("year sighting")
plt.tight_layout()
plt.show()
```

```python
sightings_per_year = datafm['year'].value_counts().sort_index()
```

```python
plt.title("Number of Mythical Sightings Per Year")
plt.bar(sightings_per_year.index.astype(str), sightings_per_year.values)
plt.xlabel("Year")
plt.ylabel("Number of Sightings")
plt.tight_layout()
plt.show()
```
