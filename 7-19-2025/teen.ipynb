```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("teen_addict.csv")
df.head()
```

```python
for x in range(min(30,len(df))):
    name=df.loc[x,"Name"]
    age=df.loc[x,"Age"]
    gender=df.loc[x,"Gender"]
    location=df.loc[x,"Location"]  
    print(name,age,gender,location)
```

```python
plt.title("teen addicts")
plt.bar(df["Gender"],df["Age"])
plt.xlabel("Gender")
plt.ylabel("Age")
plt.tight_layout()
plt.show()
```
